# Plain Python Version (No Libraries)

Formula : J(w, b) = (1 / 2m) * sum(( prediction - actual) **2)

```python
def cost_fun(x_list, y_list, w, b):
    
    m = len(x_list)
    total_err = 0

    for i in range(m):
        pred = w * x_list[i]+b
        err = pred - y_list[i]
        total_err += err ** 2

    return total_err / (2 * m)
```

```python
# the same 5-row table from Blog 2
distances = [1, 2, 3, 4, 5]
actual_times = [12, 17, 22, 27, 32]
```

```python
wrong_guess = cost_fun(distances, actual_times, w=4, b=10)
```

```python
correct_guess = cost_fun(distances, actual_times, w=5, b=7)
```

```python
print("Cost for w=4, b=10: \t", wrong_guess)
print("Cost for w=5, b=7: \t", correct_guess)
```

# Vectorized NumPy Version

```python
import numpy as np
```

```python
def cost_fun_vectorized(x, y, w, b):

    m = len(x)
    pred = w * x + b
    err = pred - y
    return np.sum(err ** 2) / (2 * m)
```

```python
distances_np = np.array([1, 2, 3, 4, 5])
actual_times_np = np.array([12, 17, 22, 27, 32])
```

```python
wrong_guess_np = cost_fun_vectorized(distances_np, actual_times_np, w=4, b=10)
correct_guess_np = cost_fun_vectorized(distances_np, actual_times_np, w=5, b=7)
```

```python
print("Cost for w=4, b=10: \t", wrong_guess_np)
print("Cost for w=5, b=7: \t", correct_guess_np)
```

# Comparing A Few Different Guesses

```python
guesses = [
    (5, 7),    # the correct fit
    (4, 10),   # a decent but wrong guess
    (3, 10),   # a worse guess
]
 
print("Comparing multiple guesses:\n")
for w, b in guesses:
    cost = cost_fun_vectorized(distances_np, actual_times_np, w, b)
    print(f"w={w}, b={b}  ->  cost = {cost}\n")
```
