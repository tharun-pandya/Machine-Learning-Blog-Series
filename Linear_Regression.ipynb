# Plain Pthon Version

```python
def predict(x, w, b):
    """
    Predicts an output using the linear regression formula.

    x : input value (e.g. distance in km)
    w : weight - how much the prediction changes per unit of x
    b : bias - the baseline prediction when x is 0
    """
    return w * x + b
```

```python
#distances we want predictions
distances = [1, 2, 3, 4, 5, 6]
```

```python
#values found by hand
w = 5
b = 7
```

```python
predictions_loop = []
for x in distances:
    predictions_loop.append(predict(x, w, b))

print("Plain loop version:")
print(predictions_loop)
```

# Vectorized NumPy version

```python
import numpy as np
```

```python
distances_np = np.array([1, 2, 3, 4, 5, 6])
```

```python
predictions_vectorized = w * distances_np + b
```

```python
print("Vectorized NumPy version:")
print(predictions_vectorized)
```
