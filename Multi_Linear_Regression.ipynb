# Predicting with a known w and b

```python
import numpy as np
```

```python
def predict(X, w, b):
    return X.dot(w) + b
```

# Gradient descent, extended to a vector of weights

```python
def gradient_descent(X, y, w, b, learning_rate, iterations, track_history=False):
    m = len(y)
    history = []
 
    for i in range(iterations):
        predictions = X.dot(w) + b
        errors = predictions - y
 
        gradient_w = X.T.dot(errors) / m
        gradient_b = np.sum(errors) / m
 
        w = w - learning_rate * gradient_w
        b = b - learning_rate * gradient_b
 
        if track_history and (i + 1) in [1, 10, 100, 1000, iterations]:
            history.append((i + 1, w.copy(), b))
 
    if track_history:
        return w, b, history
    return w, b
```

# The example table from the blog

```python
X = np.array([
    [1, 2, 0],
    [2, 1, 1],
    [3, 3, 0],
    [2, 4, 1],
    [4, 2, 0],
], dtype=float)
 
actual_times = np.array([16, 29, 28, 35, 31], dtype=float)
```

```python
w_start = np.zeros(3)
b_start = 0.0
```

```python
final_w, final_b, history = gradient_descent(
    X, actual_times, w_start, b_start,
    learning_rate=0.05, iterations=10000, track_history=True
)
```

```python
print("Final weights:", final_w)
```

```python
print("Final bias:", final_b)
```

```python
print("\nHow the weights moved during training:")
print(f"{'Iterations':>10} | {'w1':>6} | {'w2':>6} | {'w3':>6} | {'b':>6}")
for steps, w_snapshot, b_snapshot in history:
    print(f"{steps:>10} | {w_snapshot[0]:>6.2f} | {w_snapshot[1]:>6.2f} | {w_snapshot[2]:>6.2f} | {b_snapshot:>6.2f}")
```

# Verifying predictions against the actual table

```python
predictions = predict(X, final_w, final_b)
 
print("Checking predictions against the actual table:\n")
for i in range(len(actual_times)):
    print(f"Order {i+1}: predicted = {predictions[i]:.2f}, actual = {actual_times[i]}")
 
```
