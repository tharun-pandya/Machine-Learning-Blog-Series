```python
import numpy as np
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

### The three key behaviors of the curve

```python
print("Very negative z (approaches 0):")
print("sigmoid(-6) =", sigmoid(-6))

print("\nz = 0 (exactly 0.5):")
print("sigmoid(0) =", sigmoid(0))

print("\nVery positive z (approaches 1):")
print("sigmoid(6) =", sigmoid(6))
```

### The 2.73 example from the blog

```python
z = 2.73
probability = sigmoid(z)
print(f"\nRaw score z = {z}")
print(f"sigmoid({z}) = {probability:.4f}")
print(f"As a percentage: {probability * 100:.1f}%")
```

### A few more values, to see the curve take shape

```python
print("\nA spread of values across the curve:")
for z in [-10, -6, -2.73, -1, 0, 1, 2.73, 6, 10]:
    print(f"z = {z:>6} \t->\t sigmoid(z) = {sigmoid(z):.4f}")
```

### visualization

```python
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 200)
y = []
```

```python
for i in x:
    y.append(sigmoid(i))
```

```python
plt.plot(x, y, c="b")
plt.axhline(y=0.5, color='r', linestyle='--')
plt.title("The Sigmoid Function")
plt.xlabel("z")
plt.ylabel("sigmoid(z)")
plt.grid()
plt.show()
```
